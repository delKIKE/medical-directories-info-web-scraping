**Web Scraping: Hospital Viera, Hospital Policlinica, and Honduras Medical Center directories**

This notebook scrapes the aforementioned hospitals' doctor directories by using Beautiful Soup and saves each doctor's information in a CSV file for further use. For the Honduras Medical Center, since the site is a JavaScript-rendered Single Page Application (SPA), Selenium was used to open a real Chrome browser and wait for the content to load before saving the HTML. BeautifulSoup was then used to parse the HTML and extract each doctor's name, specialty, sub-specialty, floor, phone number, and schedule.

**Policlinica**

```python
print(policlinica_response.text)
```

```python
policlinica_soup = BeautifulSoup(policlinica_response.text, 'html.parser')
```

```python
policlinica_whole_list = policlinica_soup.find_all('div', class_='site-content')

with open('policlinica_directory.csv', 'w', newline='') as csvfile:
  writer= csv.writer(csvfile)

  specialty_write = ""
  write_doctor_name = ""
  write_specialties = ""
  write_telephone = ""
  write_office_hours = ""

  for entry in policlinica_whole_list:
    doctor_info = entry.find_all('div', class_= 'elementor-container elementor-column-gap-no')

    for div in doctor_info:



      #only find the SPECIALTY
      only_specialty = div.find('div', class_= 'elementor-widget-container')
      regex_specialty = r'^(Dr|Dra|directorio)'
      if not re.match(regex_specialty, only_specialty.get_text().lstrip()):
          print(only_specialty.get_text())
          specialty_write = only_specialty.get_text()




      doc_info= div.find_all('div', class_= 'elementor-widget-container')
      for p in doc_info[1:]:
        policlinica_info_to_write = p.get_text().lstrip()


        #ONLY name
        regex_doc_name = r'^(Dr|Dra)\.'
        if re.match(regex_doc_name, policlinica_info_to_write):
          print(policlinica_info_to_write)
          write_doctor_name = policlinica_info_to_write

        #doctor's specific specialties
        regex_all_specialties = r'^(.*?)(?=Tel\.)'
        specialties_match = re.search(regex_all_specialties, policlinica_info_to_write, re.DOTALL)
        if specialties_match:
          print(specialties_match.group(1))
          write_specialties = specialties_match.group(1)

        #ONLY telephone
        regex_telephone = r'\d{4}-\d{4}(?:\s*/\s*\d+|\s*al\s*\d+\s*ext\s*#?\d+)?'
        telephone_match = re.search(regex_telephone, policlinica_info_to_write)
        if telephone_match:
          print(telephone_match.group(0).strip())
          write_telephone = telephone_match.group(0).strip()

        #office hours
        regex_office_hours = r'(Lunes|Martes|Miércoles|Jueves|Viernes|Sábado|Domingo|Por cita|Por orden de llegada)[\s\S]*?(?=(\d{4}-\d{4})|$)'
        office_hours_match = re.search(regex_office_hours, policlinica_info_to_write)
        if office_hours_match:
          print(office_hours_match.group().strip())
          write_office_hours = office_hours_match.group().strip()

        writer.writerow([write_doctor_name, write_specialties, write_telephone, write_office_hours])





```

```python
with open('hmc_page.html', 'r', encoding='utf-8') as file:
  html_content = file.read()

#now parse the html content by using beautiful soup
soup = BeautifulSoup(html_content, 'html.parser')


doctor_entry = soup.find_all('li', class_= 'flex justify-between gap-x-6 py-5')


with open('medical_directory.csv', 'w', newline='') as csvfile:
  writer= csv.writer(csvfile)
  for li in doctor_entry:
    doctor_info = li.find_all('p')

    print(doctor_info[0].get_text().lstrip())
    doctor_name = doctor_info[0].get_text().lstrip()

    print(doctor_info[1].get_text().lstrip())
    specialty = doctor_info[1].get_text().lstrip()

    print(doctor_info[2].get_text().lstrip())
    floor = doctor_info[2].get_text().lstrip()

    print(doctor_info[3].get_text().lstrip())
    phones = doctor_info[3].get_text().lstrip()

    print(doctor_info[4].get_text().lstrip())
    attending= doctor_info[4].get_text().lstrip()

    print(doctor_info[5].get_text().lstrip())
    extra_row_perhaps = doctor_info[5].get_text().lstrip()


    writer.writerow([doctor_name, specialty, floor, phones, attending, extra_row_perhaps])
    print()


```
